## Problem 1 — Generators: chunked
What "chunked" means: splitting a long sequence into smaller, fixed-size groups. Example: the list [1,2,3,4,5,6,7] split into chunks of size 3 becomes [1,2,3], [4,5,6], [7] (the last group can be smaller if it doesn't divide evenly).

The problem: write a generator function chunked(iterable, size) that yields successive chunks of length size from iterable, as lists.

```python
from itertools import batched

def chunked(iterable,size):
    for i in batched(iterable,size):
        yield i

l=[1,2,3,4,5,6,7]

demo=chunked(l,3)

print(next(demo))
print(next(demo))
print(next(demo))

for chunk in chunked(l,3):
    print(chunk)
```

```python
L=[1,2,3,4,5,6,7]

def chunked(iterable,size):
    for i in range(0,len(iterable),size):
        yield iterable[i:i+size]

demo=chunked(L,3)

print(next(demo))
print(next(demo))
print(next(demo))
    
```

## Problem 2 — Decorators: count_calls
The problem: write a decorator count_calls that tracks how many times the function it decorates has been called, across all calls, and lets you check that count afterward.

```python
def count_calls(func):
    def wrapper(*args,**kwargs):
        result=func(*args,**kwargs)
        wrapper.calls+=1   # when the decorator is applied to a function, the code inside it runs and 
                          # when that function is called then the code inside the wrapper runs
        return result
    wrapper.calls=0
    return wrapper


@count_calls
def greet():
    return "hi"

greet()
greet()
greet()
print(greet.calls)
```

## Problem 4: given this list of dictionaries:

Write three separate lines of code:

Use map() to produce a list of just the names: ["Ali", "Zara", "Sam"]

Use filter() to produce a list of only the people older than 25

Use sorted() with key= to sort people by age, ascending

```python
people = [
    {"name": "Ali", "age": 34},
    {"name": "Zara", "age": 22},
    {"name": "Sam", "age": 28},
]
                
name=list(map(lambda x:x["name"],people))
print(name)
```

```python
elder=list(filter(lambda x:x["age"]>25,people))
print(elder)
```

```python
ascending_order=sorted(people,key=lambda x:x["age"])

print(ascending_order)
```

## Problem 1:
Part A:factorial

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return factorial(n-1)*n

factorial(5)
```

Part B: Sum of digits of a non negative number

```python
def digit_sum(number):
    if number<10:
        return number
    dig=number%10
    number=number//10
    return dig + digit_sum(number)
   

digit_sum(1234)
```

## Problem 2: 
write safe_divide(a, b)

```python
def safe_divide(a,b):
    try:
        division=a/b
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError:
        return "Invalid input type"
    else: # else only runs when there is no exception thrown
        print(f"Division successful: {division}")
        return division
    finally:
        print("Operation complete")

print(safe_divide(10, 2))     # works fine
print(safe_divide(10, 0))     # triggers ZeroDivisionError
print(safe_divide(10, "a"))   # triggers TypeError
```
